```python
from PIL import Image

folder = "dataset/Data/Brown spot"

corrupted = []

for file in os.listdir(folder):
    if file.lower().endswith(".jpg"):
        try:
            img = Image.open(os.path.join(folder, file))
            img.verify()
        except:
            corrupted.append(file)

print("Corrupted Images:", corrupted)
print("Total Corrupted:", len(corrupted))
```

```python
folder = "dataset/Data/Brown spot"

hashes = {}
duplicates = []

for file in os.listdir(folder):
    if file.lower().endswith(".jpg"):
        with open(os.path.join(folder, file), "rb") as f:
            file_hash = hashlib.md5(f.read()).hexdigest()

        if file_hash in hashes:
            duplicates.append(file)
        else:
            hashes[file_hash] = file

print("Duplicate Images:", duplicates)
print("Total Duplicates:", len(duplicates))
```

```python
from PIL import Image
folder = "dataset/Data/Brown spot"

ratios = set()

for file in os.listdir(folder):
    if file.lower().endswith(".jpg"):
        img = Image.open(os.path.join(folder, file))

        width, height = img.size

        ratios.add(round(width / height, 2))

print("Number of unique aspect ratios:", len(ratios))
print("Aspect Ratios:", ratios)
```
